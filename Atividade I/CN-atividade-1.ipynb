```python
import math

import numpy as np

import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
```

```python
# Parte D

def truncar(numero, casas):
    # Calcula o fator que desloca a vírgula decimal.
    fator = 10 ** casas

    # Trunca o número deslocado e devolve a vírgula à posição original.
    return math.trunc(numero * fator) / fator

def h(r, d_p, mi, L):
  Q = (math.pi * (r** 4) * d_p)/8*mi*L
  return Q

# erro_absoluto = abs(valor_referencia - valor_aproximado)

# Divide o erro absoluto pelo módulo do valor de referência.
# erro_relativo = erro_absoluto / abs(valor_referencia)

# Converte o erro relativo em porcentagem.
# erro_percentual = 100.0 * erro_relativo

r = 50 #nm
d_p = 5000  #Pa.
mi = 0.0035 #Pa s
L = 10 #μm

Q = h(r,d_p, mi,L)
```

```python
# Parte D

x1 = np.linspace(0.1, 5, 50)
x=[]
for a in x1:
  x.append(float(a))

y1 = [] #r-
y2 = [] #r+

maior_absoluto = []

y3 = [] #rel
y4 = [] #per

y5 = [] #aproximacao linear

for d_r in x:
  y1.append(abs(Q - h(r-d_r,d_p, mi,L)))
  y2.append(abs(Q - h(r+d_r,d_p, mi,L)))

for i in range(50):
  maior_absoluto.append(max(y1[i],y2[i]))

for i in range(50):
  y3.append(maior_absoluto[i]/Q)

for i in range(50):
  y4.append(100*y3[i])

for d_r in x:
  y5.append((4 * d_r) / r)

print (y5)
```
